# Rich diff 테스트 노트북
두 번째 커밋에서 이 셀을 조금 수정해 보세요.

```python
import math
import matplotlib.pyplot as plt

values = [0, 0.25, 0.5, 0.75, 1.0]
sines = [math.sin(v) for v in values]

plt.figure(figsize=(5,3))
plt.plot(values, sines, marker='o')
plt.title('Sine over sample points')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show()
```
